## 0.4 循环：C自己控制计数，Python遍历一个范围

```c

```

Python：

动手前，小林这样做：不运行代码，写出range(1, 4)产生的全部整数。然后，他按这条提示核对：range左闭右开，停止值4不包含在结果中。

## 0.7 用循环观察逐年增长

小林看着新的Python写法有点紧张，因为一个符号写错，结果就会变。前面留下了一个线索：上一节只看起点和终点，现在展开中间状态。所以，他想先弄清：**闭式公式背后的逐年更新规则是什么？**

现在把前面学过的变量、循环和输出组合起来。这里的`range(years)`产生从0到`years-1`的整数，所以显示年份时使用`year + 1`。

动手前，小林这样做：预测每年增加的元数是否相等，以及列表为何有years+1项。然后，他按这条提示核对：append保存新状态；第0项必须是初始财富。

- `wealth_history[0]`是初始财富，`wealth_history[10]`是第10年末财富。

## 0.10 把完整计算封装成函数

小林看着新的Python写法有点紧张，因为一个符号写错，结果就会变。前面留下了一个线索：把向量公式升级为有输入约定的计算工具。所以，他想先弄清：**如何阻止明显错误输入悄悄生成结论？**

输入 → 规则 → 输出。规范的代码写出的函数应准确说明单位、参数范围和假设。

动手前，小林这样做：判断years=0、rate=0、负本金分别应返回还是报错。然后，他按这条提示核对：验证输入，再计算并return；非法输入用ValueError。

输入检查保护计算流程，但是不能证明固定收益率假设真实。


````python
def future_value(principal, annual_rate, years):
    # 计算固定年收益率、按年复利时的终值。
    if principal < 0:
        raise ValueError("principal不能为负数")
    if years < 0:
        raise ValueError("years不能为负数")
    return principal * (1 + annual_rate) ** years


examples = [
    (1_000, 0.03, 0),
    (1_000, 0.00, 10),
    (1_000, 0.03, 10),
]

for p, r, n in examples:
    print(p, r, n, "->", round(future_value(p, r, n), 2))
````

### 边界条件检查

- `years=0`时终值应等于本金；
- `annual_rate=0`时财富不增长；
- 本金或年份为负时，本函数主动拒绝输入。

**Python提示：`raise`**

`raise ValueError(...)`：输入不符合本函数的约定，阻止错误数据，所以预防产生错误结论。

## 0.12 参数滑块：自己控制模型

小林看着新的Python写法有点紧张，因为一个符号写错，结果就会变。前面留下了一个线索：上一节只比较固定参数，现在主动改变假设。所以，他想先弄清：**结论对收益率和期限有多敏感？**

安装`ipywidgets`后可以拖动收益率和期限。可以在当前环境下执行python -m pip install ipywidgets安装库函数。pycharm的库函数管理可参考：https://blog.csdn.net/m0_69437736/article/details/132671714。
这里与c语言的管理方式是相近的，故不作过多阐述。

动手前，小林这样做：在拖动前判断提高收益率或延长期限会怎样改变曲线。然后，他按这条提示核对：widget只是调用绘图函数；无法显示时直接运行固定参数函数。

但是，参数实验回答“如果假设改变”，不预测真实收益。


````python
def plot_compound(rate=0.03, horizon=20):
    t = np.arange(horizon + 1)
    w = principal * (1 + rate) ** t
    fig, ax = plt.subplots()
    ax.plot(t, w, color="#2563eb", linewidth=2)
    ax.set(title=f"年收益率={rate:.1%}，期限={horizon}年",
           xlabel="年份", ylabel="财富（元）")
    plt.show()
    print({"rate": rate, "horizon": horizon, "final_wealth": round(float(w[-1]), 2)})


try:
    from ipywidgets import interact, FloatSlider, IntSlider
    interact(
        plot_compound,
        rate=FloatSlider(value=0.03, min=-0.10, max=0.15, step=0.01, description="年收益率"),
        horizon=IntSlider(value=20, min=1, max=50, step=1, description="年数"),
    )
except ImportError:
    print("未安装ipywidgets：请执行 python -m pip install ipywidgets")
    plot_compound()
````

**读图卡｜复利参数实验**

- **轴与单位**：横轴是年份，纵轴是财富（元）；标题显示当前假设的年收益率与期限。
- **观察**：拖动前先预测曲率和终值，再比较实际变化。
- **解释**：敏感性来自指数中的收益率和期限共同作用。
- **不能推出**：滑块展示模型假设，不提供真实收益概率或投资建议。
- **静态后备**：控件不显示或图不刷新时，直接运行 plot_compound(rate=0.03, horizon=20)。


## 0.13 调试：先读错误，再改代码

小林看着新的Python写法有点紧张，因为一个符号写错，结果就会变。前面留下了一个线索：模型变长后，错误定位也成为实验能力。所以，他想先弄清：**怎样从异常类型找到最小修复？**
